```python
# Connect customer activities with product details

customer_product_data = pd.merge(
    customer_data,
    product_data[
        ['product_id', 'product_name',
         'category', 'colour', 'style']
    ],
    on='product_id',
    how='left'
)

print("CUSTOMER PRODUCT INTERACTION DATA")

display(customer_product_data.head(10))
```

```python
# Apply the assignment weights; clicks are tracked as behaviour, not preference.
customer_product_data['activity_score'] = (
    customer_product_data['event_type']
    .map(PREFERENCE_WEIGHTS)
    .fillna(0.0)
)

print("WEIGHTED PREFERENCE ACTIVITY SCORES")
display(
    customer_product_data[
        ['user_id', 'product_id', 'event_type', 'activity_score']
    ]
)
```

```python
# Calculate weighted and normalized category preferences
category_preferences = customer_product_data.groupby(
    ['user_id', 'category']
)['activity_score'].sum().reset_index()
category_preferences['normalized_score'] = category_preferences.groupby(
    'user_id'
)['activity_score'].transform(lambda scores: scores / scores.sum())

favourite_category = category_preferences.loc[
    category_preferences.groupby('user_id')['activity_score'].idxmax()
].reset_index(drop=True).rename(
    columns={'category': 'Favourite_Category'}
)
print("CUSTOMER FAVOURITE CATEGORIES")
display(category_preferences)
display(favourite_category)
```

```python
# Calculate weighted and normalized colour preferences
colour_preferences = customer_product_data.groupby(
    ['user_id', 'colour']
)['activity_score'].sum().reset_index()
colour_preferences['normalized_score'] = colour_preferences.groupby(
    'user_id'
)['activity_score'].transform(lambda scores: scores / scores.sum())

favourite_colour = colour_preferences.loc[
    colour_preferences.groupby('user_id')['activity_score'].idxmax()
].reset_index(drop=True).rename(
    columns={'colour': 'Favourite_Colour'}
)
print("CUSTOMER FAVOURITE COLOURS")
display(colour_preferences)
display(favourite_colour)
```

```python
# Retain a separate colour-score display for the exploration section.
colour_preferences = customer_product_data.groupby(
    ['user_id', 'colour']
)['activity_score'].sum().reset_index()
colour_preferences['normalized_score'] = colour_preferences.groupby(
    'user_id'
)['activity_score'].transform(lambda scores: scores / scores.sum())
print("NORMALIZED CUSTOMER COLOUR PREFERENCE SCORES")
display(colour_preferences)
```

```python
# Calculate weighted and normalized style preferences
style_preferences = customer_product_data.groupby(
    ['user_id', 'style']
)['activity_score'].sum().reset_index()
style_preferences['normalized_score'] = style_preferences.groupby(
    'user_id'
)['activity_score'].transform(lambda scores: scores / scores.sum())

favourite_style = style_preferences.loc[
    style_preferences.groupby('user_id')['activity_score'].idxmax()
].reset_index(drop=True).rename(
    columns={'style': 'Favourite_Style'}
)
print("CUSTOMER FAVOURITE STYLES")
display(style_preferences)
display(favourite_style)
```

```python
# Combine customer preferences

customer_preferences = favourite_category.merge(
    favourite_colour,
    on='user_id',
    how='outer'
).merge(
    favourite_style,
    on='user_id',
    how='outer'
)

print("FINAL CUSTOMER PREFERENCES")

display(customer_preferences)
```

```python
# Standardise product categories, colours and styles

customer_product_data['category'] = (
    customer_product_data['category']
    .astype(str).str.strip().str.title()
)

customer_product_data['colour'] = (
    customer_product_data['colour']
    .astype(str).str.strip().str.title()
)

customer_product_data['style'] = (
    customer_product_data['style']
    .astype(str).str.strip().str.title()
)

print("Preference values standardised successfully!")
```

```python
# Recalculate favourite preferences

def get_favourite(column, output_name):

    scores = customer_product_data.groupby(
        ['user_id', column]
    )['activity_score'].sum().reset_index()

    scores = scores.sort_values(
        ['user_id', 'activity_score'],
        ascending=[True, False]
    )

    result = scores.drop_duplicates(
        'user_id'
    )[['user_id', column]]

    return result.rename(columns={
        column: output_name
    })


favourite_category = get_favourite(
    'category', 'Favourite_Category'
)

favourite_colour = get_favourite(
    'colour', 'Favourite_Colour'
)

favourite_style = get_favourite(
    'style', 'Favourite_Style'
)

# Combine preferences without unnecessary score columns

customer_preferences = favourite_category.merge(
    favourite_colour, on='user_id', how='outer'
).merge(
    favourite_style, on='user_id', how='outer'
)

display(customer_preferences)
```

```python
# Customer-based product recommendation

def recommend_products(user_id, top_n=5):

    # Get customer preferences
    user = customer_preferences[
        customer_preferences['user_id'] == user_id
    ]

    if user.empty:
        print("Customer not found!")
        return

    category = user['Favourite_Category'].iloc[0].lower()
    colour = user['Favourite_Colour'].iloc[0].lower()
    style = user['Favourite_Style'].iloc[0].lower()

    recommendations = product_features.copy()

    # Calculate preference matching scores
    recommendations['Category_Score'] = (
        recommendations['category'] == category
    ).astype(int)

    recommendations['Colour_Score'] = (
        recommendations['colour'] == colour
    ).astype(int)

    recommendations['Style_Score'] = (
        recommendations['style'] == style
    ).astype(int)

    # Calculate final recommendation score
    recommendations['Recommendation_Score'] = (
        recommendations['Category_Score'] * 0.4 +
        recommendations['Colour_Score'] * 0.3 +
        recommendations['Style_Score'] * 0.3
    )

    # Sort products by recommendation score and rating
    recommendations = recommendations.sort_values(
        by=['Recommendation_Score', 'rating'],
        ascending=False
    )

    print("RECOMMENDED PRODUCTS FOR:", user_id)

    display(
        recommendations[
            ['product_id', 'product_name', 'category',
             'colour', 'style', 'price', 'rating',
             'Recommendation_Score']
        ].head(top_n)
    )
```

```python
def recommend_products(user_id, top_n=5):

    user = customer_preferences[
        customer_preferences['user_id'] == user_id
    ]

    if user.empty:
        print("Customer not found!")
        return

    category = user['Favourite_Category'].iloc[0].lower()
    colour = user['Favourite_Colour'].iloc[0].lower()
    style = user['Favourite_Style'].iloc[0].lower()

    recommendations = product_features.copy()

    # Calculate preference scores
    recommendations['Category_Score'] = (
        recommendations['category'] == category
    ).astype(int)

    recommendations['Colour_Score'] = (
        recommendations['colour'] == colour
    ).astype(int)

    recommendations['Style_Score'] = (
        recommendations['style'] == style
    ).astype(int)

    # Calculate final score
    recommendations['Recommendation_Score'] = (
        recommendations['Category_Score'] * 0.4 +
        recommendations['Colour_Score'] * 0.3 +
        recommendations['Style_Score'] * 0.3
    )

    # Exclude already purchased products
    purchased_products = purchase_details[
        purchase_details['user_id'] == user_id
    ]['product_id'].unique()

    recommendations = recommendations[
        ~recommendations['product_id'].isin(purchased_products)
    ]

    # Sort recommendations
    recommendations = recommendations.sort_values(
        by=['Recommendation_Score', 'rating'],
        ascending=False
    )

    print("PERSONALIZED RECOMMENDATIONS FOR:", user_id)

    display(
        recommendations[
            ['product_id', 'product_name', 'category',
             'colour', 'style', 'price', 'rating',
             'Recommendation_Score']
        ].head(top_n)
    )
```

```python
recommend_products('U001', 5)
recommend_products('U006', 5)
recommend_products('U009', 5)
```

```python
# Generate recommendations for all customers

all_recommendations = []

for user_id in customer_preferences['user_id']:

    user = customer_preferences[
        customer_preferences['user_id'] == user_id
    ]

    category = user['Favourite_Category'].iloc[0].lower()
    colour = user['Favourite_Colour'].iloc[0].lower()
    style = user['Favourite_Style'].iloc[0].lower()

    rec = product_features.copy()

    rec['Category_Score'] = (
        rec['category'] == category
    ).astype(int)

    rec['Colour_Score'] = (
        rec['colour'] == colour
    ).astype(int)

    rec['Style_Score'] = (
        rec['style'] == style
    ).astype(int)

    rec['Recommendation_Score'] = (
        rec['Category_Score'] * 0.4 +
        rec['Colour_Score'] * 0.3 +
        rec['Style_Score'] * 0.3
    )

    purchased = purchase_details[
        purchase_details['user_id'] == user_id
    ]['product_id'].unique()

    rec = rec[
        ~rec['product_id'].isin(purchased)
    ]

    rec = rec.sort_values(
        by=['Recommendation_Score', 'rating'],
        ascending=False
    ).head(5)

    rec['user_id'] = user_id

    all_recommendations.append(rec)

all_recommendations = pd.concat(
    all_recommendations,
    ignore_index=True
)

print("ALL CUSTOMER RECOMMENDATIONS")
display(all_recommendations)
```

## Final customer feature outputs

Run the code cell above to generate `outputs/user_features.parquet`, `outputs/customer_features.csv`, and the customer spending CSVs. The feature builder uses purchase events only for RFM and spending, and view (1), wishlist (2), add-to-cart (3), and purchase (5) events for weighted preference vectors. Clicks are counted for behavior and engagement, but are not assigned a preference weight because the project specification does not define one.

RFM scores range from 1 to 5; lower recency is better. Segments are rules-based and should be treated as illustrative with this small sample. Spending is catalogue list price multiplied by purchased quantity; discounts are not treated as actual transaction discounts. Purchases with missing catalogue prices are flagged and their total spend is left unavailable. Material and room type are calculated because those columns exist in the product catalogue; the builder also handles their absence without making up values. Wishlist activity is not in the current behavior data, so its counts are zero.

The final CSV and Parquet contain the same customer-level features. Preference-vector column names retain the `category_pref__`, `colour_pref__`, and `style_pref__` prefixes used by the recommender.
