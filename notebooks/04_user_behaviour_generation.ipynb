
# User Behaviour Generation

## Project: AI-Based Furniture Recommendation System

**Role:** Person 2 – User Behaviour Generation

### Objective
Generate synthetic user behaviour data for a furniture recommendation system, including:
- Product views
- Add-to-cart activities
- Wishlist activities
- Purchase activities

The generated behaviour data is linked to valid product IDs from the furniture catalogue.


```python
# Import required libraries

from pathlib import Path
import pandas as pd
import sys

# Set project directory
BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

# Define file paths
PRODUCT_FILE = BASE_DIR / "data" / "raw" / "products.csv"
BEHAVIOUR_FILE = BASE_DIR / "data" / "raw" / "user_behavior.csv"

print("Project directory:", BASE_DIR)
print("Product file exists:", PRODUCT_FILE.exists())
print("Behaviour file exists:", BEHAVIOUR_FILE.exists())
```

```python
# Load product and user behaviour datasets

products = pd.read_csv(PRODUCT_FILE)
behaviour = pd.read_csv(BEHAVIOUR_FILE)

print("Product Dataset Shape:", products.shape)
print("User Behaviour Dataset Shape:", behaviour.shape)

print("\nFirst 5 User Behaviour Records:")
display(behaviour.head())
```

```python
# Count each type of user activity

event_counts = behaviour["event_type"].value_counts()

print("User Activity Distribution:")
display(event_counts)
```

```python
# Validate user behaviour dataset

unique_users = behaviour["user_id"].nunique()

purchase_counts = (
    behaviour[behaviour["event_type"] == "purchase"]
    .groupby("user_id").size()
)

valid_product_ids = set(products["product_id"])
invalid_product_events = (
    ~behaviour["product_id"].isin(valid_product_ids)
).sum()

print("Unique users:", unique_users)
print("Minimum purchases per user:", purchase_counts.min())
print("Maximum purchases per user:", purchase_counts.max())
print("Invalid product IDs:", invalid_product_events)

assert unique_users >= 2000
assert purchase_counts.min() >= 1
assert purchase_counts.max() <= 10
assert invalid_product_events == 0

print("\nAll validation checks passed!")
```
