# UrbanNest Furnishings – Data Generation

## Objective
Generate a synthetic furniture catalogue containing 600
unique furniture products.

## Product Attributes
- Product ID
- Product Name
- Category
- Material
- Colour
- Style
- Price
- Rating
- Dimensions
- Room Type
- Description

```python
import sys
from pathlib import Path

BASE_DIR = Path.cwd()

if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

sys.path.insert(0, str(BASE_DIR))

from src.product_generator import (
    generate_products,
    validate_products,
    save_products
)

print("Product generator imported successfully!")
```

```python
products = generate_products(600)

print("Total products generated:", len(products))

display(products.head())
```

```python
validate_products(products)

print("\nUnique IDs:", products["product_id"].nunique())
print("Missing values:", products.isnull().sum().sum())
print("Duplicate IDs:", products["product_id"].duplicated().sum())
```

```python
save_products(products)

print("\nProduct catalogue generated successfully!")
```

```python
print("DATASET SUMMARY")

print("Total products:", len(products))

print("\nCategory distribution:")
display(products["category"].value_counts())

print("\nMaterial distribution:")
display(products["material"].value_counts())

print("\nAverage price:", round(products["price"].mean(), 2))

print("\nAverage rating:", round(products["rating"].mean(), 2))
```

```python
print("Material Distribution")
display(products["material"].value_counts())

print("\nColour Distribution")
display(products["colour"].value_counts())
```

```python
print("Price and Rating Analysis")

print("Average Price:", round(products["price"].mean(), 2))
print("Minimum Price:", products["price"].min())
print("Maximum Price:", products["price"].max())

print("Average Rating:", round(products["rating"].mean(), 2))
print("Minimum Rating:", products["rating"].min())
print("Maximum Rating:", products["rating"].max())
```
